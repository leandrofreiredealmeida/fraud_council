## 3. Análise da distribuição das principais features

## Distribuição das Classes

```python
NORD_PALETTE = [
    "#88C0D0", "#A3BE8C", "#EBCB8B",
    "#BF616A", "#B48EAD", "#D08770", "#81A1C1",
]

plt.rcParams.update({
    "figure.facecolor": "#2E3440",
    "axes.facecolor": "#3B4252",
    "text.color": "#ECEFF4",
    "axes.labelcolor": "#ECEFF4",
    "xtick.color": "#D8DEE9",
    "ytick.color": "#D8DEE9",
    "axes.edgecolor": "#4C566A",
    "grid.color": "#434C5E",
})

images_dir = Path("../images")
images_dir.mkdir(exist_ok=True)

class_counts = df["Class"].value_counts().sort_index()
labels = ["Legítima (0)", "Fraude (1)"]
colors = [NORD_PALETTE[0], NORD_PALETTE[3]]

fig, ax = plt.subplots(figsize=(7, 5))
bars = ax.bar(labels, class_counts.values, color=colors, edgecolor="#4C566A", linewidth=0.8)

for bar, count in zip(bars, class_counts.values):
    ax.text(
        bar.get_x() + bar.get_width() / 2,
        bar.get_height() + 2000,
        f"{count:,}\n({count / len(df) * 100:.2f}%)",
        ha="center", va="bottom", fontsize=11, color="#ECEFF4",
    )

ax.set_title("Distribuição das Classes", fontsize=14, pad=15)
ax.set_ylabel("Número de Transações", fontsize=11)
ax.set_ylim(0, class_counts.max() * 1.15)
for spine in ["top", "right"]:
    ax.spines[spine].set_visible(False)

plt.tight_layout()
plt.savefig(images_dir / "class_distribution.png", dpi=150, bbox_inches="tight")
plt.show()
```

## Distribuição do Valor das Transações (Amount)

```python
fig, axes = plt.subplots(1, 2, figsize=(14, 5))

ax = axes[0]
ax.hist(df["Amount"], bins=100, color=NORD_PALETTE[0], edgecolor="#4C566A", linewidth=0.3, alpha=0.9)
ax.set_yscale("log")
ax.set_title("Distribuição de Amount (escala log)")
ax.set_xlabel("Amount (USD)")
ax.set_ylabel("Frequência (log)")
for spine in ["top", "right"]:
    ax.spines[spine].set_visible(False)

ax = axes[1]
for cls, color, label in zip([0, 1], [NORD_PALETTE[0], NORD_PALETTE[3]], ["Legítima", "Fraude"]):
    ax.hist(
        df[df["Class"] == cls]["Amount"],
        bins=80,
        color=color,
        alpha=0.65,
        label=label,
        edgecolor="#4C566A",
        linewidth=0.3,
    )
ax.set_xlim(0, 500)
ax.set_title("Amount por Classe (até USD 500)")
ax.set_xlabel("Amount (USD)")
ax.set_ylabel("Frequência")
ax.legend(facecolor="#3B4252", edgecolor="#4C566A", labelcolor="#ECEFF4")
for spine in ["top", "right"]:
    ax.spines[spine].set_visible(False)

fig.suptitle("Distribuição do Valor das Transações", fontsize=15, y=1.02)
plt.tight_layout()
plt.savefig(images_dir / "amount_distribution.png", dpi=150, bbox_inches="tight")
plt.show()
```

## Distribuição Temporal (Time)

```python
fig, ax = plt.subplots(figsize=(12, 5))

ax.hist(
    df[df["Class"] == 0]["Time"] / 3600,
    bins=96,
    color=NORD_PALETTE[0],
    alpha=0.8,
    label="Legítima",
    edgecolor="#4C566A",
    linewidth=0.2,
)
ax.hist(
    df[df["Class"] == 1]["Time"] / 3600,
    bins=96,
    color=NORD_PALETTE[3],
    alpha=0.9,
    label="Fraude",
    edgecolor="#4C566A",
    linewidth=0.2,
)

ax.set_title("Distribuição Temporal das Transações", fontsize=14, pad=12)
ax.set_xlabel("Tempo (horas desde o início do registro)", fontsize=11)
ax.set_ylabel("Número de Transações", fontsize=11)
ax.legend(facecolor="#3B4252", edgecolor="#4C566A", labelcolor="#ECEFF4")
for spine in ["top", "right"]:
    ax.spines[spine].set_visible(False)

plt.tight_layout()
plt.savefig(images_dir / "time_distribution.png", dpi=150, bbox_inches="tight")
plt.show()
```

## Distribuição das Features PCA (V1–V28)

```python
v_cols = [f"V{i}" for i in range(1, 29)]
fraud = df[df["Class"] == 1]
legit = df[df["Class"] == 0].sample(n=10_000, random_state=42)

fig, axes = plt.subplots(7, 4, figsize=(18, 22))
axes_flat = axes.flatten()

for i, col in enumerate(v_cols):
    ax = axes_flat[i]
    ax.hist(legit[col], bins=60, color=NORD_PALETTE[0], alpha=0.6, density=True, label="Legítima", edgecolor="none")
    ax.hist(fraud[col], bins=60, color=NORD_PALETTE[3], alpha=0.7, density=True, label="Fraude", edgecolor="none")
    ax.set_title(col, fontsize=11)
    ax.tick_params(labelsize=8)
    for spine in ["top", "right"]:
        ax.spines[spine].set_visible(False)
    if i == 0:
        ax.legend(facecolor="#3B4252", edgecolor="#4C566A", labelcolor="#ECEFF4", fontsize=8)

fig.suptitle("Distribuição das Features PCA (V1–V28) por Classe", fontsize=16, y=1.005)
plt.tight_layout()
plt.savefig(images_dir / "pca_features_distribution.png", dpi=120, bbox_inches="tight")
plt.show()
```

**Conclusões**
- Observamos que apenas 0.17% das transações foram fraudes, o que indica que devemos ter cuidado com qualquer análise devido à possibilidade de um conjunto de dados "skewed" (distorcido).
- A maior parte das transações são de valores baixos (até 300 USD).
- Observamos também que o nosso conjunto de dados é restrito a apenas 48 horas e no gráfico da distribuição temporal é possível observar a subida e queda da quantidade de transações por causa do períodos diurnos e noturnos.
- No gráfico de distribuição das features é importante notar as features mais discriminativas para modelos (V3, V4, V11, V12, V14) daquelas com sobreposição completa ou quase compelta que oferecem pouca informação para modelos (V19, V20, V22, V24, V25).
